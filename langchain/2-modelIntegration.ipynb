### GROQ Model Integration

```python
from langchain_groq import ChatGroq

model = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

model
```

```python
## invoke the model
response = model.invoke("write a poem about the ocean")
response.content
```
